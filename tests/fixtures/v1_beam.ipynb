```python
from handcalcs.decorator import handcalc

@handcalc(override="long", precision=2)
def moment(w, L):
    """Moment of a simple span."""
    # a note for readers only
    ## Bending moment
    M = w * L**2 / 8
    return M

latex, M_max = moment(5, 6)
print(latex)
```

```python
@handcalc(jupyter_display=True)
def shear(w, L):
    V = w * L / 2
    return V

V_max = shear(5, 6)
```
